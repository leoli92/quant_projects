```python
import requests
import pandas as pd

API_KEY = "6dZ3ouRdPCS58jEdYI7KPPynm_Rd88Qn"

def get_option_contracts(underlying: str) -> pd.DataFrame:
    url = "https://api.massive.com/v3/reference/options/contracts"
    params = {
        "underlying_ticker": underlying,
        "apiKey": API_KEY,
        "limit": 1000,
    }

    rows = []
    while True:
        r = requests.get(url, params=params, timeout=30)
        r.raise_for_status()
        data = r.json()

        rows.extend(data.get("results", []))

        next_url = data.get("next_url")
        if not next_url:
            break

        url = next_url
        params = {"apiKey": API_KEY}

    df = pd.DataFrame(rows)
    return df

contracts = get_option_contracts("SPY")
print(contracts.head())
print(contracts.columns.tolist())
print(len(contracts))
```
